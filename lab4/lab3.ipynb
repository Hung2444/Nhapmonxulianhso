```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Đường dẫn ảnh
input_path = "exercise/colorful-ripe-tropical-fruits.jpg"
output_path = "exercise/kiwi_translated.jpg"

# Đọc ảnh gốc
image = cv2.imread(input_path)

if image is None:
    print("Không thể đọc ảnh từ đường dẫn:", input_path)
    exit()

# Tọa độ vùng quả kiwi (set cứng)
# Định dạng: (x, y, width, height)
kiwi_region = (220, 150, 100, 100)  # Tọa độ cứng của quả kiwi

x, y, w, h = kiwi_region

# Cắt phần ảnh chứa quả kiwi
kiwi = image[y:y+h, x:x+w]

# Tịnh tiến quả kiwi sang phải 30 pixels
translated_image = image.copy()
translated_image[y:y+h, x+30:x+30+w] = kiwi

# Hiển thị kết quả bằng matplotlib
fig, axs = plt.subplots(1, 2, figsize=(10, 5))
axs[0].imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
axs[0].set_title("Ảnh Gốc")
axs[0].axis("off")

axs[1].imshow(cv2.cvtColor(translated_image, cv2.COLOR_BGR2RGB))
axs[1].set_title("Ảnh Sau Khi Tịnh Tiến Kiwi")
axs[1].axis("off")

plt.tight_layout()

# Lưu ảnh kết quả
cv2.imwrite(output_path, translated_image)

plt.show()
```

```python
import imageio.v2 as iio
import numpy as np
import matplotlib.pyplot as plt
import cv2


img = iio.imread('exercise/colorful-ripe-tropical-fruits.jpg')
img = cv2.cvtColor(img, cv2.COLOR_RGB2BGR)


y1, y2 = 340, 790
x1, x2 = 100, 650
papaya = img[y1:y2, x1:x2]
papaya_hsv = cv2.cvtColor(papaya, cv2.COLOR_BGR2HSV)
papaya_hsv[:, :, 0] = 60 
img[y1:y2, x1:x2] = cv2.cvtColor(papaya_hsv, cv2.COLOR_HSV2BGR)


y1, y2 = 300, 1000
x1, x2 = 1660, 3500
melon = img[y1:y2, x1:x2]
melon_hsv = cv2.cvtColor(melon, cv2.COLOR_BGR2HSV)
melon_hsv[:, :, 0] = 120  
img[y1:y2, x1:x2] = cv2.cvtColor(melon_hsv, cv2.COLOR_HSV2BGR)


img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)


plt.imshow(img_rgb)
plt.axis('off')
plt.title("Đổi màu đu đủ và dưa hấu")
plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Đường dẫn ảnh
input_path = "exercise/pagoda.jpg"
output_path = "exercise/pagoda_resized.jpg"

# Đọc ảnh gốc
image = cv2.imread(input_path)

# Kiểm tra nếu ảnh không thể đọc được
if image is None:
    print("Không thể đọc ảnh từ đường dẫn:", input_path)
    exit()

# Tọa độ vùng ngôi chùa (giả định tọa độ)
# Định dạng: (x, y, width, height)
pagoda_region = (100, 100, 200, 300)  # Tọa độ ngôi chùa

# Cắt vùng ngôi chùa
x, y, w, h = pagoda_region
pagoda = image[y:y+h, x:x+w]

# Tăng kích thước ngôi chùa lên 5 lần
pagoda_resized = cv2.resize(pagoda, (w * 5, h * 5), interpolation=cv2.INTER_CUBIC)

# Tạo ảnh kết quả (để hiển thị rõ ràng ngôi chùa đã tăng kích thước)
output_image = np.full((h * 5, w * 5, 3), 255, dtype=np.uint8)  # Tạo ảnh nền trắng
output_image[:pagoda_resized.shape[0], :pagoda_resized.shape[1]] = pagoda_resized

# Hiển thị kết quả bằng matplotlib
fig, axs = plt.subplots(1, 2, figsize=(12, 6))
axs[0].imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
axs[0].set_title("Ảnh Gốc")
axs[0].axis("off")

axs[1].imshow(cv2.cvtColor(output_image, cv2.COLOR_BGR2RGB))
axs[1].set_title("Ngôi Chùa Sau Khi Phóng To (5 Lần)")
axs[1].axis("off")

plt.tight_layout()

# Lưu ảnh kết quả
cv2.imwrite(output_path, output_image)

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Đường dẫn đến các hình trong thư mục exercise
images = {
    "1": "exercise/colorful-ripe-tropical-fruits.jpg",
    "2": "exercise/quang_ninh.jpg",
    "3": "exercise/pagoda.jpg"
}

def load_image(image_path):
    image = cv2.imread(image_path)
    if image is None:
        print(f"Không thể đọc ảnh từ {image_path}")
        return None
    return image

def translate(image, tx, ty):
    height, width = image.shape[:2]
    matrix = np.float32([[1, 0, tx], [0, 1, ty]])
    return cv2.warpAffine(image, matrix, (width, height))

def rotate(image, angle):
    height, width = image.shape[:2]
    center = (width // 2, height // 2)
    matrix = cv2.getRotationMatrix2D(center, angle, 1.0)
    return cv2.warpAffine(image, matrix, (width, height))

def scale_image(image, scale_x, scale_y):
    return cv2.resize(image, None, fx=scale_x, fy=scale_y, interpolation=cv2.INTER_CUBIC)

def coordinate_map(image):
    plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
    plt.title("Coordinate Map")
    plt.show()

def display_image(original, result, title):
    # Hiển thị ảnh gốc và ảnh kết quả
    fig, axes = plt.subplots(1, 2, figsize=(10, 5))
    axes[0].imshow(cv2.cvtColor(original, cv2.COLOR_BGR2RGB))
    axes[0].set_title("Ảnh Gốc")
    axes[0].axis("off")

    axes[1].imshow(cv2.cvtColor(result, cv2.COLOR_BGR2RGB))
    axes[1].set_title(title)
    axes[1].axis("off")

    plt.tight_layout()
    plt.show(block=True)  # Đảm bảo dừng lại để xem ảnh
    plt.close(fig)  # Đóng cửa sổ sau khi hiển thị

def main():
    while True:
        # Hiển thị menu
        print("\nMenu:")
        print("T - Tịnh tiến")
        print("X - Xoay")
        print("P - Phóng to")
        print("H - Thu nhỏ")
        print("C - Coordinate Map")
        print("Q - Thoát")

        choice = input("Chọn một phép biến đổi: ").strip().upper()

        if choice == "Q":
            print("Thoát chương trình.")
            break

        # Chọn hình để thực hiện phép biến đổi
        print("\nHình ảnh có sẵn:")
        for key, path in images.items():
            print(f"{key}: {path}")

        image_choice = input("Chọn hình (1, 2, 3): ").strip()
        if image_choice not in images:
            print("Lựa chọn không hợp lệ!")
            continue

        image = load_image(images[image_choice])
        if image is None:
            continue

        # Thực hiện phép biến đổi
        if choice == "T":  # Tịnh tiến
            tx = int(input("Nhập giá trị tịnh tiến theo trục X: "))
            ty = int(input("Nhập giá trị tịnh tiến theo trục Y: "))
            result = translate(image, tx, ty)
            title = f"Tịnh tiến ({tx}, {ty})"
        elif choice == "X":  # Xoay
            angle = float(input("Nhập góc xoay (độ): "))
            result = rotate(image, angle)
            title = f"Xoay {angle}°"
        elif choice == "P":  # Phóng to
            scale = float(input("Nhập hệ số phóng to (ví dụ: 2.0): "))
            result = scale_image(image, scale, scale)
            title = f"Phóng to {scale} lần"
        elif choice == "H":  # Thu nhỏ
            scale = float(input("Nhập hệ số thu nhỏ (ví dụ: 0.5): "))
            result = scale_image(image, scale, scale)
            title = f"Thu nhỏ {scale} lần"
        elif choice == "C":  # Coordinate Map
            coordinate_map(image)
            continue
        else:
            print("Lựa chọn không hợp lệ!")
            continue

        # Hiển thị và lưu kết quả
        display_image(image, result, title)
        output_path = f"exercise/result_{choice}.jpg"
        cv2.imwrite(output_path, result)
        print(f"Đã lưu kết quả vào {output_path}")

if __name__ == "__main__":
    main()
```
